1. Remove Duplicates from Sorted Array
 | Two-pointer (in-place) | 
 [LC 26](https://leetcode.com/problems/remove-duplicates-from-sorted-array/)

```python
def removeDuplicates( nums):
        if not nums:
            return 0
        slow=0
        for fast in range(1,len(nums)):
            if nums[fast]!=nums[slow]:
                slow+=1
                nums[slow]=nums[fast]
        return slow+1   
removeDuplicates([0,0,0,1,1,2,2,])        
```

6. Maximum Subarray (Kadane's) | Kadane's algorithm | [LC 53]

```python
def maxSubArray(nums):
        current_sum=nums[0]
        max_sum=nums[0]
        for i in range(1,len(nums)):
            current_sum=max(nums[i],nums[i]+current_sum)
            max_sum=max(current_sum,max_sum)
        return max_sum    
maxSubArray([-2,1,-3,4,-1,2,1,-5,4])        
```

9. Merge Sorted Array | Two-pointer from end | [LC 88](https://leetcode.com/problems/merge-sorted-array/) |

```python
def merge( nums1, m, nums2, n):
        
        p1 = m - 1
        p2 = n - 1
        p_write = m + n - 1

        
        while p1 >= 0 and p2 >= 0:
            if nums1[p1] > nums2[p2]:
                nums1[p_write] = nums1[p1]
                p1 -= 1
            else:
                nums1[p_write] = nums2[p2]
                p2 -= 1
            p_write -= 1

        
        while p2 >= 0:
            nums1[p_write] = nums2[p2]
            p2 -= 1
            p_write -= 1
        return nums1    
nums1=[1,2,3,0,0,0]
m=3
nums2=[2,5,6]
n=3
merge(nums1,3,nums2,3)            
```

15. Pascal's Triangle | DP / Row building | [LC 118](https://leetcode.com/problems/pascals-triangle/) |


```python

def generate( numRows):
        triangle = [[1]]  # row 0 is always [1]

        for i in range(1, numRows):
            row = []
            prev_row = triangle[i-1]
            for j in range(i + 1):
                if j == 0 or j == i:
                    row.append(1)
                else:
                    row.append(prev_row[j-1] + prev_row[j])
            triangle.append(row)

        return triangle
generate(3)        
```

16. Find Pivot Index | Prefix sum | [LC 724](https://leetcode.com/problems/find-pivot-index/) |

```python
def pivotIndex(arr):
    total_sum=sum(arr)
    left_sum=0

    for num in range(len(arr)):
        total_sum-=arr[num]

        if left_sum==total_sum:
            return num
        left_sum+=arr[num]
    return -1
arr=[1,7,3,6,5,6]
pivotIndex(arr)            
```

20. Longest Common Prefix | Horizontal/Vertical scanning | [LC 14](https://leetcode.com/problems/longest-common-prefix/) |

```python
def longestCommonPrefix(strs):
    shortest_string=min(strs,key=lambda x: len(x))

    for i,char in enumerate(shortest_string):
        if not all(word[i]==char for word in strs):
            return shortest_string[:i]
    return shortest_string
longestCommonPrefix(['flower','flow','flight'])            
```

25. Add Binary | String traversal from end | [LC 67](https://leetcode.com/problems/add-binary/) |

```python
class Solution(object):
    def addBinary(self, a, b):
        i, j = len(a) - 1, len(b) - 1
        carry = 0
        result = []
        
        while i >= 0 or j >= 0 or carry:
            bit_a = int(a[i]) if i >= 0 else 0
            bit_b = int(b[j]) if j >= 0 else 0
            
            total = bit_a + bit_b + carry
            result.append(str(total % 2))
            carry = total // 2
            
            i -= 1
            j -= 1
        
        return ''.join(reversed(result))
soln=Solution()
soln.addBinary('11','1')    
```

27.  Reverse Vowels of a String | Two-pointer | [LC 345](https://leetcode.com/problems/reverse-vowels-of-a-string/) |

```python
class Solution:
    def reverseVowels(self, s):
        left = 0
        right = len(s) - 1
        vowels = set('aeiouAEIOU')
        s = list(s)

        while left < right:
            if s[left] not in vowels:
                left += 1
            elif s[right] not in vowels:
                right -= 1
            else:
                s[left], s[right] = s[right], s[left]
                left += 1
                right -= 1

        return ''.join(s)
s='IceCreAm'
soln=Solution()
soln.reverseVowels(s)   
```

28. Squares of a Sorted Array | Two-pointer from ends | [LC 977](https://leetcode.com/problems/squares-of-a-sorted-array/) |

```python
class Solution(object):
    def sortedSquares(self, nums):
        left = 0
        right = len(nums) - 1
        result = [0] * len(nums)
        pos = len(nums) - 1

        while left <= right:
            if nums[left]**2 > nums[right]**2:
                result[pos] = nums[left]**2
                left += 1
            else:
                result[pos] = nums[right]**2
                right -= 1
            pos -= 1

        return result
soln=Solution()
soln.sortedSquares([-4,-1,0,3,10])    
```

```python
class Solution(object):
    def validPalindrome(self, s):
        def is_palindrome(s,l,r):
          while l<r:   
            if s[l]!=s[r]:
                return False
            l+=1
            r-=1
          return True
        left,right=0,len(s)-1
        while left<=right:
            if s[left]==s[right]:
                left+=1
                right-=1
            else:
                return is_palindrome(s,left+1,right) or is_palindrome(s,left,right-1)
        return True             
                    
s='aba'
soln=Solution()
soln.validPalindrome(s)        
```

```python
def twoSum(numbers, target):
    left, right = 0, len(numbers)-1
    
    while left < right:
        current_sum = numbers[left]+numbers[right]        
        if current_sum == target:
            return left+1,right+1
        elif current_sum < target:
            left+=1
        else:
            right-=1
numbers = [2,3,4] 
target = 6            
twoSum(numbers,target)
```

31. Merge Strings Alternately | Two-pointer | [LC 1768](https://leetcode.com/problems/merge-strings-alternately/) |

```python
a='ab'
s='pqrs'
def mergeAlternately(word1: str, word2: str) -> str:
    i, j = 0, 0
    output = []
    
    while i < len(word1) and j < len(word2):
        output.append(word1[i])
        output.append(word2[j])
        i += 1
        j += 1
    
    # one of these will be empty (i or j already exhausted), the other has the leftover
    output.extend(word1[i:])
    output.extend(word2[j:])
    
    return ''.join(output)
mergeAlternately(a,s)
```

33. Backspace String Compare | Two-pointer / Stack | [LC 844](https://leetcode.com/problems/backspace-string-compare/) |

```python
class Solution:
    def build(self, string):
        stack = []
        for ch in string:
            if ch.islower():
                stack.append(ch)
            else:
                if stack:
                    stack.pop()
        return ''.join(stack)
    
    def backspaceCompare(self, s, t):

        return self.build(s) == self.build(t)

soln=Solution() 
soln.backspaceCompare('ab#c','ad#c')   
```

```python
nums=[2,4,1,3]
class Solution:
    def sortArrayByParity(self, nums):
        left = 0
        right = len(nums) - 1
        while left < right:
            if nums[left] % 2 == 0:
                left += 1
            elif nums[right] % 2 != 0:
                right -= 1
            else:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
                right -= 1
        return nums
soln=Solution()
soln.sortArrayByParity(nums)           
        
```

35. Container With Most Water | Two-pointer (area max) | [LC 11](https://leetcode.com/problems/container-with-most-water/) |

```python
class Solution:
    def maxArea(self, height):
        left = 0
        right = len(height) - 1
        best = 0
        
        while left < right:
            width = right - left
            limiting_height = min(height[left], height[right])
            area = width * limiting_height
            best = max(best, area)
            
            if height[left] < height[right]:
                left += 1
            else:
                right -= 1
        
        return best
soln=Solution()    
nums=[1,8,6,2,5,4,8,3,7]
soln.maxArea(nums)
    

```

```python
class Solution:
    def groupAnagrams(self, strs):
        groups = {}
        for word in strs:
            signature = ''.join(sorted(word))
            if signature not in groups:
                groups[signature] = []
            groups[signature].append(word)
        return list(groups.values())
soln=Solution()
soln.groupAnagrams(["eat","tea","tan","ate","nat","bat"] )    
```

```python
class Solution(object):
    def containsNearbyDuplicate(self, nums, k):
        seen={}
        for i,num in enumerate(nums):
            if num in seen:
                last_index=seen[num]
                if i-last_index<=k:
                    return True
            seen[num]=i
        return False  
soln=Solution()
soln.containsNearbyDuplicate([1,2,3,1],3)    
```

40. Word Pattern | HashMap (bijection) | [LC 290](https://leetcode.com/problems/word-pattern/) |

```python
class Solution(object):
    def wordPattern(self, pattern, s):
        words = s.split()
        if len(pattern) != len(words):
            return False
        
        map_pattern_to_word = {}
        map_word_to_pattern = {}
        
        for i in range(len(pattern)):
            char_pattern = pattern[i]
            word = words[i]

            if char_pattern in map_pattern_to_word:
                if map_pattern_to_word[char_pattern] != word:
                    return False
            else:
                map_pattern_to_word[char_pattern] = word

            if word in map_word_to_pattern:
                if map_word_to_pattern[word] != char_pattern:
                    return False
            else:
                map_word_to_pattern[word] = char_pattern
                
        return True
soln=Solution()
soln.wordPattern('abba','dog cat cat dog')    
```

43. Two Sum IV - Input is BST | HashMap + DFS | [LC 653](https://leetcode.com/problems/two-sum-iv-input-is-a-bst/) |

```python
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution(object):
    def findTarget(self, root, k):
        seen = set()
        
        def find_pair(node):
            if node is None:
                return False
            if k - node.val in seen:
                return True
            seen.add(node.val)
            return find_pair(node.left) or find_pair(node.right)
        
        return find_pair(root)


# ---- Example usage ----
# Building this tree:
#         5
#        / \
#       3   6
#      / \    \
#     2   4    7

root = TreeNode(5)
root.left = TreeNode(3)
root.right = TreeNode(6)
root.left.left = TreeNode(2)
root.left.right = TreeNode(4)
root.right.right = TreeNode(7)

sol = Solution()

print(sol.findTarget(root, 9))   # Example 1 -> True   (5 + 4 = 9)
print(sol.findTarget(root, 28))  # Example 2 -> False  (max possible sum is 6+7=13, nowhere near 28)
```
