Q1. A can complete a task in 18 days, B can complete the same task in 15 days. B works alone for 10 days and then quits. How many more days does A need to finish the rest, working alone?
- a) 7
- b) 6
- c) 5
- d) 10

Here is the **smallest, simplest explanation** possible, without any extra steps:

---

## 🧠 Super Simple Explanation

**Step 1:** Assume the total work is **90 units** (LCM of 18 and 15, just to avoid fractions).

**Step 2:**
- **A** does = 90 ÷ 18 = **5 units per day**
- **B** does = 90 ÷ 15 = **6 units per day**

**Step 3:**  
B works for 10 days → B does = 6 × 10 = **60 units** of work.

**Step 4:**  
Work left = 90 − 60 = **30 units**.

**Step 5:**  
A does 5 units per day, so time needed = 30 ÷ 5 = **6 days**.

---

## ✅ Final Answer: **6 days** (Option b)

**Q2.** The HCF of 2472, 1284, and a third number N is 12. Their LCM is 2³ × 3² × 5 × 103 × 107. What is N?
- a) 2² × 3² × 7

# 📘 Q2 - Complete Solution (Step-by-Step)

---

## 📝 Question:
> The HCF of **2472**, **1284**, and a third number **N** is **12**.  
> Their LCM is **2³ × 3² × 5 × 103 × 107**.  
> What is **N**?

**Options:**  
a) 2² × 3² × 7  
b) 2² × 3³ × 103  
c) 2² × 3 × 5  
d) None of these

---

## 🧮 Step-by-Step Solution

---

### Step 1: Prime Factorize 2472 and 1284

**2472**  
2472 ÷ 2 = 1236  
1236 ÷ 2 = 618  
618 ÷ 2 = 309  
309 ÷ 3 = 103 (103 is prime, stop here)

**So, 2472 = 2 × 2 × 2 × 3 × 103 = 2³ × 3 × 103** ✅

---

**1284**  
1284 ÷ 2 = 642  
642 ÷ 2 = 321  
321 ÷ 3 = 107 (107 is prime, stop here)

**So, 1284 = 2 × 2 × 3 × 107 = 2² × 3 × 107** ✅

---

### Step 2: Write What We Know

| Number | Prime Factorization |
|--------|---------------------|
| 2472 | 2³ × 3 × 103 |
| 1284 | 2² × 3 × 107 |
| N | ? |

**Given:**  
HCF = 12 = **2² × 3**  
LCM = **2³ × 3² × 5 × 103 × 107**

---

### Step 3: Find What N Must Have (From LCM)

Look at the LCM and see what's **missing** from 2472 and 1284:

| Factor | In 2472 | In 1284 | In LCM | Who gives it? |
|--------|---------|---------|--------|---------------|
| 2³ | ✅ Yes | ❌ No (only 2²) | ✅ Yes | 2472 gives 2³ |
| 3¹ | ✅ Yes | ✅ Yes | — | Both have it |
| 3² | ❌ No | ❌ No | ✅ Yes | **N must give 3²** |
| 5 | ❌ No | ❌ No | ✅ Yes | **N must give 5** |
| 103 | ✅ Yes | ❌ No | ✅ Yes | 2472 gives 103 |
| 107 | ❌ No | ✅ Yes | ✅ Yes | 1284 gives 107 |

**So from LCM, N must have: 3² × 5**

---

### Step 4: Find What N Must Have (From HCF)

HCF = 12 = **2² × 3**

HCF takes the **smallest** power of each factor among ALL numbers.

| Factor | In 2472 | In 1284 | What can N have? |
|--------|---------|---------|------------------|
| 2 | Has 2³ | Has 2² | N can have **2²** (not 2³, or HCF would become 2³) |
| 3 | Has 3¹ | Has 3¹ | N can have 3² (2472 and 1284 keep HCF at 3¹) |

**So from HCF, N must have: 2²**

---

### Step 5: Combine Everything

From Step 3: N must have **3² × 5**  
From Step 4: N must have **2²**

**Therefore, N = 2² × 3² × 5**

---

### Step 6: Check the Options

| Option | Value | Matches? |
|--------|-------|----------|
| a) 2² × 3² × 7 | Has 7 instead of 5 | ❌ No |
| b) 2² × 3³ × 103 | Has 3³ (too high) and 103 | ❌ No |
| c) 2² × 3 × 5 | Missing one 3 (only 3¹, need 3²) | ❌ No |
| d) None of these | — | ✅ Yes |

---

## ✅ Final Answer:

**(d) None of these**

Because the correct value of N is **2² × 3² × 5**, which is not listed in the options.
